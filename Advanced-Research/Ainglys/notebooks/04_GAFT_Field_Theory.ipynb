# GAFT: Granular Arithmonic Field Theory

This notebook demonstrates the GAFT framework:
- **Sub-symbolic field computation** using particle interactions
- **Arithmonic granules** as fundamental computing units
- **Field dynamics** with energy conservation

```python
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.insert(0, '/home/runner/workspace/Advanced-Research/Ideas/projects/gaft/src')

from gaft import (
    ArithmonicGranule,
    GranuleType,
    ArithmonicField,
    GranularSimulation,
)

print("GAFT imported successfully!")
```

## 1. Arithmonic Granules

```python
g1 = ArithmonicGranule(
    position=np.array([0.0, 0.0]),
    momentum=np.array([1.0, 0.5]),
    phase=0.0,
    granule_type=GranuleType.ACTIVE,
)

g2 = ArithmonicGranule(
    position=np.array([1.0, 1.0]),
    momentum=np.array([-0.5, 0.0]),
    phase=np.pi/4,
    granule_type=GranuleType.PASSIVE,
)

print(f"Granule 1: position={g1.position}, energy={g1.energy:.4f}")
print(f"Granule 2: position={g2.position}, energy={g2.energy:.4f}")
```

## 2. Arithmonic Field

```python
field = ArithmonicField(
    dimensions=2,
    coupling_strength=0.5,
    decay_rate=0.1,
)

field.add_granule(g1)
field.add_granule(g2)

print(f"Field has {field.num_granules} granules")
print(f"Field total energy: {field.total_energy:.4f}")
```

```python
# Compute field at grid points
x = np.linspace(-2, 2, 50)
y = np.linspace(-2, 2, 50)
X, Y = np.meshgrid(x, y)

field_values = field.compute_field(X, Y)

# Visualize the field
fig, ax = plt.subplots(figsize=(10, 8))
contour = ax.contourf(X, Y, field_values, levels=20, cmap='viridis')
plt.colorbar(contour, label='Field Value')

for g in field.granules:
    ax.plot(g.position[0], g.position[1], 'ro', markersize=10)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Arithmonic Field with Granules')
plt.tight_layout()
plt.show()
```

## 3. Granular Simulation

```python
np.random.seed(42)

n_granules = 20
initial_granules = []

for i in range(n_granules):
    pos = np.random.randn(2) * 0.5
    mom = np.random.randn(2) * 0.1
    g = ArithmonicGranule(
        position=pos,
        momentum=mom,
        phase=np.random.uniform(0, 2*np.pi),
        granule_type=GranuleType.ACTIVE,
    )
    initial_granules.append(g)

simulation = GranularSimulation(
    initial_granules,
    coupling_strength=0.3,
    dt=0.1,
)

print(f"Simulation initialized with {n_granules} granules")
print(f"Initial total energy: {simulation.total_energy:.4f}")
```

```python
# Run simulation
n_steps = 100
energy_history = [simulation.total_energy]

for step in range(n_steps):
    simulation.step()
    energy_history.append(simulation.total_energy)

print(f"Final total energy: {simulation.total_energy:.4f}")
print(f"Energy change: {energy_history[-1] - energy_history[0]:.4f}")
```

```python
# Plot energy evolution
plt.figure(figsize=(10, 4))
plt.plot(energy_history)
plt.xlabel('Simulation Step')
plt.ylabel('Total Energy')
plt.title('Energy Conservation in Granular Simulation')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
