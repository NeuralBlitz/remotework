# GAAV: Granular Arithmetic and Visualization

This notebook demonstrates the GAAV framework:
- **Granular arithmetic** G = (v, delta, sigma)
- **Uncertainty propagation** through algebraic operations
- **Visualization tools** for granular data

```python
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.insert(0, '/home/runner/workspace/Advanced-Research/Ideas/projects/gaav/src')

from gaav import (
    Granule,
    UncertaintyType,
    GranularOperators,
    GranularVisualizer,
)

print("GAAV imported successfully!")
```

## 1. Creating Granules

```python
# Create granules with different uncertainty types
g1 = Granule(value=10.0, delta=0.5, sigma=0.2)
g2 = Granule(value=5.0, delta=0.3, sigma=0.1)
g3 = Granule(value=8.0, uncertainty_type=UncertaintyType.SYSTEMATIC)
g4 = Granule(value=12.0, uncertainty_type=UncertaintyType.STOCHASTIC)

print(f"g1 = {g1}")
print(f"g2 = {g2}")
print(f"g3 = {g3}")
print(f"g4 = {g4}")
```

```python
# Granule properties
print(f"g1 value: {g1.value}")
print(f"g1 delta: {g1.delta}")
print(f"g1 sigma: {g1.sigma}")
print(f"g1 total uncertainty: {g1.total_uncertainty}")
```

## 2. Granular Arithmetic Operations

```python
ops = GranularOperators()

g_sum = ops.add(g1, g2)
g_diff = ops.subtract(g1, g2)
g_prod = ops.multiply(g1, g2)
g_quot = ops.divide(g1, g2)

print(f"g1 + g2 = {g_sum}")
print(f"g1 - g2 = {g_diff}")
print(f"g1 * g2 = {g_prod}")
print(f"g1 / g2 = {g_quot}")
```

## 3. Uncertainty Propagation Visualization

```python
x_vals = np.linspace(1, 10, 50)
granules = [Granule(x, x*0.1, x*0.05) for x in x_vals]

sq_values = [ops.multiply(g, g).value for g in granules]
sq_unc = [ops.multiply(g, g).total_uncertainty for g in granules]

exp_values = [ops.exp(g).value for g in granules]
exp_unc = [ops.exp(g).total_uncertainty for g in granules]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(x_vals, sq_values, 'b-', lw=2)
axes[0].fill_between(x_vals, np.array(sq_values)-np.array(sq_unc), np.array(sq_values)+np.array(sq_unc), alpha=0.3)
axes[0].set_xlabel('x')
axes[0].set_ylabel('x^2')
axes[0].set_title('Uncertainty in x^2')
axes[0].grid(True, alpha=0.3)

axes[1].plot(x_vals, exp_values, 'r-', lw=2)
axes[1].fill_between(x_vals, np.array(exp_values)-np.array(exp_unc), np.array(exp_values)+np.array(exp_unc), alpha=0.3)
axes[1].set_xlabel('x')
axes[1].set_ylabel('exp(x)')
axes[1].set_title('Uncertainty in exp(x)')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
